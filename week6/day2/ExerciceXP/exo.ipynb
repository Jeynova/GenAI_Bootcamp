```python
import warnings
warnings.filterwarnings("ignore")

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
import matplotlib.pyplot as plt

print("\n" + "="*100)
print("Exercise 1 — What are Large Language Models (LLMs)?")
print("="*100)

# 1) Q1 — Explain what LLMs are (in plain words)
ex1_answer = """
Large Language Models (LLMs) are very large neural networks trained on huge amounts of text to learn statistical
regularities of language. They can understand, complete, transform, summarize, and generate text. They don’t “think”
like humans, but they are extremely good at modeling the probability of the next token (word/piece of word) given a
context. Because of this, you can prompt them with instructions or examples and they will generate coherent, useful
continuations.
"""
print(ex1_answer)

# 2) Load a pretrained model & tokenizer (GPT‑2)
print("\nLoading GPT-2… (Exercise 1 – step 3)")
model_name = "gpt2"  # (you could swap it for another causal LM)
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
print(f"Model '{model_name}' loaded successfully!")
print("""
GPT‑2 is a causal language model, meaning it predicts the next token in a sequence.
It was trained on a large, diverse corpus and can generate coherent, contextually relevant text.
""")

print("\n" + "="*100)
print("Exercise 2 — Transformer Architecture & Tokenization")
print("="*100)

# 1) Explain tokenization + choose a sentence
ex2_answer = """
Tokenization is the process of converting raw text into smaller units (tokens) that the model can understand.
With GPT‑2 / byte-level BPE tokenizers, tokens can be whole words or subword pieces. Special prefixes (like 'Ġ') mark
spaces (word boundaries). The tokenizer then maps tokens to numeric IDs so the model can process them as tensors.
"""
print(ex2_answer)

text = "Transformers tokenize text into subword units to handle rare words efficiently."
# 2) Tokenize the input text
tokens = tokenizer.tokenize(text)
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print(f"Original Text: {text}")
print(f"Tokens: {tokens}")
print(f"Token IDs: {token_ids}")

# 3) Visualize the tokenization
plt.figure(figsize=(12, 4))
plt.bar(range(len(token_ids)), token_ids, color="skyblue")
plt.xticks(range(len(tokens)), tokens, rotation=90)
plt.xlabel("Tokens")
plt.ylabel("Token IDs")
plt.title("Tokenization with GPT-2 (byte-level BPE)")
plt.tight_layout()
plt.show()

print("\n" + "="*100)
print("Exercise 3 — Understanding Token IDs and Special Prefixes")
print("="*100)

# Print each token and its ID
for t, i in zip(tokens, token_ids):
    print(f"{t:15s} -> {i}")

ex3_answer = """
About the 'Ġ' prefix:
In GPT‑2 / RoBERTa-style byte-level BPE tokenizers, 'Ġ' marks that the token starts with a whitespace in the original text.
It’s a way for the tokenizer to remember word boundaries without using spaces as actual tokens.
"""
print(ex3_answer)

print("\n" + "="*100)
print("Exercise 4 — Pretraining vs. Fine-Tuning")
print("="*100)

ex4_answer = """
Pretraining:
- The model is trained on a huge, generic text corpus with a self-supervised objective (e.g., predicting the next token).
- It learns general language patterns, grammar, facts, and broad skills.

Fine-Tuning:
- The pretrained model is adapted to a specific downstream task or dataset (e.g., sentiment analysis, QA).
- The objective changes (e.g., classification cross-entropy), and only a fraction of data is needed.
- Fine-tuning “specializes” the general capabilities learned during pretraining.
"""
print(ex4_answer)

print("\n" + "="*100)
print("Exercise 5 — Generate Simple Text")
print("="*100)

# Prompt
input_text = "In the future, artificial intelligence will"
inputs = tokenizer.encode(input_text, return_tensors="pt")

# Text generation (next-token prediction)
torch.manual_seed(42)
output_ids = model.generate(
    inputs,
    max_new_tokens=50,
    do_sample=True,
    top_k=50,
    top_p=0.95,
    temperature=0.9,
    eos_token_id=tokenizer.eos_token_id,
)
output_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)

print(f"Input: {input_text}")
print(f"\nGenerated Output:\n{output_text}")
```
