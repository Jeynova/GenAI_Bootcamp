```python
from transformers import BertTokenizerFast

tokenizer = BertTokenizerFast.from_pretrained("bert-base-uncased")

sentence = "Natural Language Processing with BERT is powerful."

# Tokenisation “brute”
tokens = tokenizer.tokenize(sentence)
print("Tokens sans spéciaux :", tokens)

# Encodage complet pour le modèle (avec [CLS]/[SEP], padding, truncation)
enc = tokenizer(
    sentence,
    padding="max_length",
    truncation=True,
    max_length=16,
    return_tensors="pt",
    return_token_type_ids=True,
    return_attention_mask=True
)

print("\nInput IDs :", enc["input_ids"])
print("Tokens    :", tokenizer.convert_ids_to_tokens(enc["input_ids"][0]))
print("Attention :", enc["attention_mask"])
print("TokenType :", enc["token_type_ids"])

# Repérer les tokens spéciaux
ids = enc["input_ids"][0].tolist()
specials = [(i, tokenizer.convert_ids_to_tokens([tok_id])[0]) for i, tok_id in enumerate(ids) if tok_id in [tokenizer.cls_token_id, tokenizer.sep_token_id, tokenizer.pad_token_id]]
print("\nTokens spéciaux (index, token) :", specials)
```

```python
import torch
import torch.nn.functional as F
from transformers import AutoTokenizer, AutoModelForSequenceClassification

class BERTSentimentAnalyzer:
    def __init__(self, model_name="distilbert-base-uncased-finetuned-sst-2-english", device=None):
        self.device = device or ("cpu")
        self.tokenizer = AutoTokenizer.from_pretrained(model_name)
        self.model = AutoModelForSequenceClassification.from_pretrained(model_name).to(self.device)
        self.model.eval()
        self.id2label = self.model.config.id2label

    def preprocess(self, text, max_length=256):
        return self.tokenizer(
            text,
            padding=True,
            truncation=True,
            max_length=max_length,
            return_tensors="pt"
        ).to(self.device)

    @torch.no_grad()
    def predict(self, text):
        enc = self.preprocess(text)
        logits = self.model(**enc).logits
        probs = F.softmax(logits, dim=-1).cpu().numpy()[0]
        pred_id = probs.argmax()
        return {
            "label": self.id2label[pred_id],
            "score": float(probs[pred_id]),
            "probs": {self.id2label[i]: float(p) for i, p in enumerate(probs)}
        }

analyzer = BERTSentimentAnalyzer()
print(analyzer.predict("This movie was terrible."))
print(analyzer.predict("This movie was outstanding!"))
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForTokenClassification

class BERTNamedEntityRecognizer:
    def __init__(self, model_name="dslim/bert-base-NER", device=None):
        self.device = device or ("cpu")
        self.tokenizer = AutoTokenizer.from_pretrained(model_name)
        self.model = AutoModelForTokenClassification.from_pretrained(model_name).to(self.device)
        self.id2label = self.model.config.id2label
        self.model.eval()

    @torch.no_grad()
    def recognize(self, text, max_length=256):
        enc = self.tokenizer(text, return_tensors="pt", truncation=True, max_length=max_length).to(self.device)
        outputs = self.model(**enc).logits  # (1, seq_len, num_labels)
        preds = outputs.argmax(dim=-1).squeeze(0).cpu().tolist()
        tokens = self.tokenizer.convert_ids_to_tokens(enc["input_ids"].squeeze(0))
        entities = []
        for tok, label_id in zip(tokens, preds):
            label = self.id2label[label_id]
            if tok.startswith("##"):  # morceau de mot, à regrouper si besoin
                tok = tok[2:]
            entities.append((tok, label))
        return entities

ner = BERTNamedEntityRecognizer()
print(ner.recognize("Apple is looking at buying U.K. startup for $1 billion"))
```

Comparaison BERT vs GPT (sans tableau)

Architecture

    BERT : encodeur (Transformers côté encoder).

    GPT : décodeur auto‑régressif (Transformers côté decoder).

Directionnalité

    BERT : bidirectionnel (Masked Language Modeling).

    GPT : unidirectionnel (left‑to‑right).

Objectif d’entraînement

    BERT : Masked Language Modeling (MLM) + Next Sentence Prediction (optionnel).

    GPT : Causal Language Modeling (CLM).

Usage principal

    BERT : compréhension, classification, NER, QA extractif.

    GPT : génération de texte, complétion, agents conversationnels.

Points forts

    BERT : représentations contextuelles riches, excellent sur les tâches de compréhension.

    GPT : très bonne génération, zéro‑shot/few‑shot, grande flexibilité des prompts.

Points faibles

    BERT : pas conçu pour générer du texte long.

    GPT : moins performant “out of the box” sur certaines tâches de matching/entailment sans ajustements.

Rôle typique dans un RAG

    BERT : encodeur d’embeddings (retriever dense, bi‑encodeur).

    GPT : génération finale à partir des documents récupérés.

Coût à l’inférence

    BERT : rapide pour encoder de gros volumes de documents (ex. Sentence‑BERT).

    GPT : plus coûteux pour la génération.

BERT dans un système RAG

Idée générale :

    BERT (ou Sentence-BERT) sert de retriever : il encode les documents et les requêtes en vecteurs (embeddings).

    Ces embeddings sont indexés dans un vector store (ex. FAISS).

    À chaque question, on récupère les k documents les plus proches (similarité cosinus / L2).

    Un modèle génératif (GPT/Flan‑T5, etc.) synthétise la réponse à partir de ces documents.

Pourquoi BERT ?

    Les modèles sentence-transformers (basés sur BERT) produisent des embeddings sémantiques robustes pour le matching requête-document.

    Ils sont légers et rapides à encoder de gros corpus.
