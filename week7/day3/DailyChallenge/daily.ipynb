**Consignes :**
1. Calcul du BLEU entre un texte de référence et une génération.
2. Calcul du ROUGE pour une autre paire de phrases.
3. Discussion sur les limites.

```python
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from rouge_score import rouge_scorer
from bert_score import score as bert_score
from huggingface_hub import login
from dotenv import load_dotenv
import os

load_dotenv()

HF_TOKEN = os.getenv("HF_TOKEN")

# ⚠️ REMPLACE par ton token ici
login(HF_TOKEN)
```

```python
# BLEU
ref_bleu = "Despite the increasing reliance on artificial intelligence in various industries, human oversight remains essential to ensure ethical and effective implementation."
gen_bleu = "Although AI is being used more in industries, human supervision is still necessary for ethical and effective application."
bleu = sentence_bleu([ref_bleu.split()], gen_bleu.split(), smoothing_function=SmoothingFunction().method1)
print(f"BLEU Score: {bleu:.4f}")

# ROUGE
ref_rouge = "In the face of rapid climate change, global initiatives must focus on reducing carbon emissions and developing sustainable energy sources to mitigate environmental impact."
gen_rouge = "To counteract climate change, worldwide efforts should aim to lower carbon emissions and enhance renewable energy development."
scorer = rouge_scorer.RougeScorer(['rouge1', 'rougeL'], use_stemmer=True)
scores = scorer.score(ref_rouge, gen_rouge)

print(f"ROUGE-1 F1: {scores['rouge1'].fmeasure:.4f}")
print(f"ROUGE-L F1: {scores['rougeL'].fmeasure:.4f}")
```

**Consignes :**
- Noter la fluidité de la réponse :  
  "Apologies, but comprehend I do not. Could you rephrase your question?"
- Proposer une version améliorée.

```python
# Likert Evaluation Simulation

response = "Apologies, but comprehend I do not. Could you rephrase your question?"

def evaluate_likert(text):
    score = 5
    if "comprehend I do not" in text:
        score -= 2
    if "Apologies" in text:
        score -= 1
    return max(score, 1)

likert_score = evaluate_likert(response)

print("Original:", response)
print(f"Likert Score: {likert_score}/5")
print("I'm sorry, I didn't quite understand. Could you please rephrase your question?")
```

Exo6

**Tâche : Résumé automatique**

**Métriques comparées :** BLEU, ROUGE, BERTScore, Perplexity, Evaluation humaine

```python
# Texte original (non utilisé ici mais contexte)
text_original = """
The global climate crisis is accelerating due to increasing carbon emissions and deforestation.
Immediate and coordinated international efforts are needed to promote renewable energy, protect forests, and reduce pollution.
"""

# Résumé généré par un LLM
summary_gen = "Global warming demands urgent action to promote clean energy and stop deforestation worldwide."

# Résumé de référence (humain)
summary_ref = "To address climate change, the world must reduce emissions, preserve forests, and support renewable energy."

# BLEU
bleu = sentence_bleu([summary_ref.split()], summary_gen.split(), smoothing_function=SmoothingFunction().method1)

# ROUGE
scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)
rouge = scorer.score(summary_ref, summary_gen)

# BERTScore
P, R, F1 = bert_score(
    [summary_gen],
    [summary_ref],
    lang="en",
    model_type="distilbert-base-uncased",
    verbose=False,
    device="cpu"
)


print(f"BLEU: {bleu:.4f}")
print(f"ROUGE-1 F1: {rouge['rouge1'].fmeasure:.4f}")
print(f"ROUGE-L F1: {rouge['rougeL'].fmeasure:.4f}")
print("Valeur de P : {:.4f}".format(P[0]), "Valeur de R : {:.4f}".format(R[0]), "Valeur de F1 : {:.4f}".format(F1[0]))
```

```python
print("Comparaison des métriques pour le résumé automatique :\n")

print("- BLEU : n-gram match, précis mais rigide.")
print("- ROUGE : capte le rappel, utile pour résumés extraits.")
print("- BERTScore : sémantique, capte le sens même avec paraphrases.")
print("- Perplexity : utile pendant l'entraînement, mais ne juge pas la qualité humaine.")
print("- Evaluation humaine : fluency, factualité, pertinence.")

print("\nRecommandation :")
print("- Utiliser ROUGE + BERTScore pour l'automatique.")
print("- Ajouter un panel humain pour vérifier factualité et clarté.")
```



- **BLEU** est relativement faible (~0.4), car la phrase générée utilise des mots différents (ex: "global warming" vs "climate change")
- **ROUGE** capture mieux l’overlap lexical général, mais reste limité en sémantique
- **BERTScore** (~0.91) montre que la phrase est **très proche en sens**, même avec une formulation différente

**Conclusion :**  
Pour la tâche de résumé, **BERTScore** est plus robuste que BLEU ou ROUGE si on souhaite mesurer la **qualité sémantique** globale.

L’évaluation humaine est recommandée en complément pour vérifier la **factualité** et la **pertinence contextuelle**.
