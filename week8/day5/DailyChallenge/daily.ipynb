LangGraph applications are built around a graph structure. As the developer, you define an application graph that models the state transitions for your application. Your app will define a state schema, and an instance of that schema is propagated through the graph.

Each node in the graph represents an action or step that can be taken. Nodes will make changes to the state in some way through code that you define. These changes can be the result of invoking an LLM, by calling an API, or executing any logic that the node defines.

Each edge in the graph represents a transition between states, defining the flow of the program. Edge transitions can be fixed, for example if you define a text-only chatbot where output is always displayed to a user, you may always transition from chatbot -> user. The transitions can also be conditional, allowing you to add branching (like an if-else statement) or looping (like for or while loops).

Now add the new tool to the graph. The get_menu tool is wrapped in a ToolNode that handles calling the tool and passing the response as a message through the graph. The tools are also bound to the llm object so that the underlying model knows they exist. As you now have a different llm object to invoke, you need to update the chatbot node so that it is aware of the tools.


from langgraph.prebuilt import ToolNode


# Define the tools and create a "tools" node.
tools = [get_menu]
tool_node = ToolNode(tools)

# Attach the tools to the model so that it knows what it can call.
llm_with_tools = llm.bind_tools(tools)


def maybe_route_to_tools(state: OrderState) -> Literal["tools", "human"]:
    """Route between human or tool nodes, depending if a tool call is made."""
    if not (msgs := state.get("messages", [])):
        raise ValueError(f"No messages found when parsing state: {state}")

    # Only route based on the last message.
    msg = msgs[-1]

    # When the chatbot returns tool_calls, route to the "tools" node.
    if hasattr(msg, "tool_calls") and len(msg.tool_calls) > 0:
        return "tools"
    else:
        return "human"

def chatbot_with_tools(state: OrderState) -> OrderState:
    """The chatbot with tools. A simple wrapper around the model's own chat interface."""
    defaults = {"order": [], "finished": False}

    if state["messages"]:
        new_output = llm_with_tools.invoke([BARISTABOT_SYSINT] + state["messages"])
    else:
        new_output = AIMessage(content=WELCOME_MSG)

    # Set up some defaults if not already set, then pass through the provided state,
    # overriding only the "messages" field.
    return defaults | state | {"messages": [new_output]}


graph_builder = StateGraph(OrderState)

# Add the nodes, including the new tool_node.

graph_builder.add_node("human", human_node)

# Human may go back to chatbot, or exit.
graph_builder.add_conditional_edges("human", maybe_exit_human_node)

# Tools always route back to chat afterwards.
graph_builder.add_edge("tools", "chatbot")

graph_builder.add_edge(START, "chatbot")

```python
class OrderState(TypedDict):
    messages: Annotated[list, add_messages]
    order: list[str]
    finished: bool

BARISTABOT_SYSINT = (
    "system",
    "Vous êtes BaristaBot, un assistant café interactif. "
    "Prenez les commandes, répondez sur les produits du menu, confirmez chaque commande avec l'utilisateur. "
    "Utilisez les fonctions : add_to_order, clear_order, get_order, confirm_order, place_order. "
    "Confirmez toujours l'ordre avec le client avant d'envoyer à la cuisine. Utilisez seulement les items/modifs du MENU. "
    "Quand la commande est terminée, remerciez et dites au revoir !"
)
WELCOME_MSG = "Bienvenue chez BaristaBot Café. Tapez `q` pour quitter. Que désirez-vous aujourd'hui ?"
```

definnissions du noeud principal du chatbot, qui prend l état actuel, ajoute l instruction système et l historique des messages, puis interroge le modèle de langage pour générer une réponse.
La construction du graphe commence ici : on crée un graphe d états avec un seul noeud chatbot, en définissant le point de départ (START) qui mène vers ce noeud.
C est une structure minimale pour lancer le bot et gérer un tour de chat, ce qui facilite les tests unitaires et la compréhension du flot principal.

```python
def chatbot(state: OrderState) -> OrderState:
    """Noeud chatbot : gère un tour de chat."""
    message_history = [BARISTABOT_SYSINT] + state["messages"]
    output = llm.invoke(message_history)
    return {"messages": [output]}

# Construction du graph
graph_builder = StateGraph(OrderState)
graph_builder.add_node("chatbot", chatbot)
graph_builder.add_edge(START, "chatbot")
chat_graph = graph_builder.compile()
```

 interaction humaine dans le flux du chatbot.

La fonction human_node affiche la dernière réponse du bot et attend la saisie de l utilisateur. Si un mot-clé d arrêt est détecté, le flag finished passe à True pour signaler la fin de la commande.

La fonction chatbot_with_welcome_msg gère le message de bienvenue lors du tout premier tour, puis le comportement normal du bot pour les échanges suivants.

La structure du graphe relie le noeud chatbot au noeud human, puis inversement, permettant une boucle conversationnelle classique.
Ce design sépare clairement les rôles humain et chatbot, rendant le flux lisible et facile à maintenir. Il anticipe aussi une gestion robuste des scénarios de sortie.

```python
def human_node(state: OrderState) -> OrderState:
    last_msg = state["messages"][-1]
    print("BaristaBot :", last_msg.content)
    user_input = input("Vous : ")
    if user_input.lower() in {"q", "quit", "exit","au revoir", "goodbye", "bye", "stop", "finish", "end", "terminer", "arrêter", "finir", "terminer la commande", "arrêter la commande", "esc"}:
        state["finished"] = True
    return state | {"messages": [("user", user_input)]}

def chatbot_with_welcome_msg(state: OrderState) -> OrderState:
    if state["messages"]:
        new_output = llm.invoke([BARISTABOT_SYSINT] + state["messages"])
    else:
        new_output = AIMessage(content=WELCOME_MSG)
    return state | {"messages": [new_output]}

graph_builder = StateGraph(OrderState)
graph_builder.add_node("chatbot", chatbot_with_welcome_msg)
graph_builder.add_node("human", human_node)
graph_builder.add_edge(START, "chatbot")
graph_builder.add_edge("chatbot", "human")
```

Ce bloc implémente la logique métier derrière chaque outil, en traitant les appels d'outils générés par le LLM (présents dans tool_msg.tool_calls).

Chaque appel (tool_call) est traité selon son nom : ajout à la commande, confirmation, affichage, annulation, ou envoi.

Les modifications à la commande sont faites sur la liste order, qui est stockée et renvoyée dans le nouvel état.

```python


def order_node(state: OrderState) -> OrderState:
    tool_msg = state["messages"][-1]
    order = state.get("order", [])
    outbound_msgs = []
    order_placed = False

    for tool_call in tool_msg.tool_calls:
        if tool_call["name"] == "add_to_order":
            modifiers = tool_call["args"].get("modifiers", [])
            modif_str = ", ".join(modifiers) if modifiers else "sans modificateurs"
            order.append(f'{tool_call["args"]["drink"]} ({modif_str})')
            response = "\n".join(order)
        elif tool_call["name"] == "confirm_order":
            print("Votre commande :")
            for drink in order: print(f"  {drink}")
            response = input("Est-ce correct ? ")
        elif tool_call["name"] == "get_order":
            response = "\n".join(order) if order else "(commande vide)"
        elif tool_call["name"] == "clear_order":
            order.clear(); response = None
        elif tool_call["name"] == "place_order":
            print("Commande envoyée !")
            print("\n".join(order))
            order_placed = True
            response = randint(1, 5)
        else:
            raise NotImplementedError(tool_call["name"])
        outbound_msgs.append(
            ToolMessage(
                content=response, name=tool_call["name"], tool_call_id=tool_call["id"]
            )
        )
    return {"messages": outbound_msgs, "order": order, "finished": order_placed}
```

Assemblage toute la logique du dialogue dans un graphe à états, permettant d'orchestrer le chatbot, les interactions humaines et l'exécution des outils.

Les outils « auto_tools » (get_menu) et « order_tools » (gestion de commande) sont tous liés au LLM pour permettre l'appel direct depuis le modèle.

Les nœuds ajoutés couvrent toutes les briques de la logique :

chatbot : gère les réponses LLM (et message de bienvenue).

human : récupère l'entrée utilisateur, permet de quitter à tout moment.

tools : exécute automatiquement les outils déclarés dans ToolNode (get_menu).

ordering : traite la logique métier propre à la commande (ajout, validation, etc.).

La fonction maybe_route_to_tools fait office de routeur dynamique après chaque réponse du LLM :

Si l'état finished est vrai, on termine la session.

Si le dernier message contient des tool_calls reconnus par ToolNode, on passe par le nœud tools.

Sinon, si tool_call non reconnu, direction ordering (logique personnalisée).

Par défaut, retour à l'humain.

Les transitions conditionnelles permettent d'enchaîner intelligemment les étapes du dialogue selon l'état courant, sans code spaghetti.

Le graphe ainsi construit est modulaire, facilement extensible, et prêt à intégrer de nouveaux outils ou briques métier sans toucher à l'architecture générale.

La compilation finale (graph_with_order_tools = graph_builder.compile()) produit un moteur conversationnel complet, interactif et multi-outils.

```python
auto_tools = [get_menu]
tool_node = ToolNode(auto_tools)
order_tools = [add_to_order, confirm_order, get_order, clear_order, place_order]
llm_with_tools = llm.bind_tools(auto_tools + order_tools)

from langgraph.graph import StateGraph, START, END

graph_builder = StateGraph(OrderState)
graph_builder.add_node("chatbot", chatbot_with_welcome_msg)
graph_builder.add_node("human", human_node)
graph_builder.add_node("tools", tool_node)
graph_builder.add_node("ordering", order_node)

def maybe_route_to_tools(state: OrderState) -> str:
    if not (msgs := state.get("messages", [])):
        raise ValueError(f"No messages found when parsing state: {state}")
    msg = msgs[-1]
    if state.get("finished", False):
        return END
    elif hasattr(msg, "tool_calls") and len(msg.tool_calls) > 0:
        if any(tool["name"] in tool_node.tools_by_name for tool in msg.tool_calls):
            return "tools"
        else:
            return "ordering"
    else:
        return "human"

graph_builder.add_conditional_edges("chatbot", maybe_route_to_tools)
graph_builder.add_conditional_edges("human", maybe_exit_human_node)
graph_builder.add_edge("tools", "chatbot")
graph_builder.add_edge("ordering", "chatbot")
graph_builder.add_edge(START, "chatbot")
graph_with_order_tools = graph_builder.compile()
```
